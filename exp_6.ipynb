```python
import pandas as pd
dates = pd.to_datetime([
    '2026-01-01',
    '2026-01-02',
    '2026-01-03',
    '2026-01-04',
    '2026-01-05'
])
temperature = [33, 34, 44, 38, 35]
time_series = pd.Series(
    temperature,
    index=dates
)
print("Time Series:")
print(time_series)
```

```python
import pandas as pd
date_index = pd.date_range(
    start='2026-01-01',
    periods=7,
    freq='D'
)
print("Generated DatetimeIndex:")
print(date_index)
```

```python
india_dates = pd.date_range(
    start='2026-01-01 09:00',
    periods=3,
    freq='h',
    tz='Asia/Kolkata'
)
print("1. Data Range with Asia/Kolkata Time Zone:")
print(india_dates)
```

```python
dates = pd.date_range(
    start='2026-01-01 09:00',
    periods=3,
    freq='h'
)
print("\n2. TimeZone-naive Data Range:")
print(dates)
```

```python
localized_dates = dates.tz_localize('Asia/Kolkata')
print("\n3. After Localizing to Asia/Kolkata:")
print(localized_dates)
```

```python
new_york_dates = localized_dates.tz_convert(
    'America/New_York'
)
print("\n4. Converted to America/New_York:")
print(new_york_dates)
```

```python
india_series = pd.Series(
    [100, 200, 300],
    index=localized_dates
)
new_york_series = pd.Series(
    [400, 500, 600],
    index=new_york_dates
)
print("\n5. India Time Series:")
print(india_series)
print("\nNew York Time Series:")
print(new_york_series)
```

```python
combined_series = pd.concat([
    india_series,
    new_york_series
])
print("\n6. Combined Time Series:")
print(combined_series)
```

```python
import pandas as pd
p=pd.Period('2026-01', freq='M')
print("Original Period:")
print(p)
print("\nAfter adding 2:")
print(p + 2)
print("\nAfter adding 5:")
print(p + 5)
print("\nAfter subtarcting 1:")
print(p - 1)
print("\nAfter subtarcting 3:")
print(p - 3)
periods = pd.period_range(
    start='2026-01',
    periods=6,
    freq='M'
)
print("\nRange of Periods:")
print(periods)

```

```python
import pandas as pd
p=pd.Period('2026-01', freq='M')
print("Original Period:")
print(p)
```

```python
daily_start = p.asfreq('D', how='start')
print("\nMonthly Period converted to Daily (Start):")
print(daily_start)
daily_end = p.asfreq('D', how='end')
print("\nMonthly Period converted to Daily (End):")
print(daily_end)
#
```
